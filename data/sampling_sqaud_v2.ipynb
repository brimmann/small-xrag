```python
import sys
from datasets import load_dataset, DatasetDict
```

```python
dataset = load_dataset("GEM/squad_v2", split='train')
```

```python
shuffled_dataset = dataset.shuffle(seed=42)
```

```python
filtered_dataset = shuffled_dataset.filter(
        lambda example: len(example['context'].split()) < 200
)
```

```python
num_to_select = 1500
sampled_dataset = filtered_dataset.select(range(num_to_select))
```

```python
train_dataset = sampled_dataset.select(range(1000))
dev_dataset = sampled_dataset.select(range(1000, 1250))
test_dataset = sampled_dataset.select(range(1250, 1500))
```

```python
final_splits = DatasetDict({
        'train': train_dataset,
        'dev': dev_dataset,
        'test': test_dataset
    })
```

```python
final_splits.push_to_hub("brimmann2/squad-v2-sampled")
```
