```python
from transformers import AutoTokenizer
from imported_code.modeling_xmistral import XMistralForCausalLM
from datasets import load_from_disk
```

```python
import torch
```

```python
XRAG_TOKEN = "<xRAG>" 
```

```python
device = "cuda" if torch.cuda.is_available() else "cpu"
```

```python
model_name = "Hannibal046/xrag-7b"
model = XMistralForCausalLM.from_pretrained(
    model_name,
    torch_dtype = torch.bfloat16,
    low_cpu_mem_usage = True,
    )
model.to(device)
tokenizer = AutoTokenizer.from_pretrained(
    model_name,
    add_eos_token=False,
    use_fast=False,
    padding_side='left'
    )
model.set_xrag_token_id(tokenizer.convert_tokens_to_ids(XRAG_TOKEN))
```

```python
rag_template = """[INST] Background: {xrag_token}, which also means: [/INST]"""
prompt = rag_template.format_map(dict(xrag_token=XRAG_TOKEN))
print(prompt)
```

```python
tokenized_prompt = tokenizer(prompt, return_tensors="pt").to(device)
```

```python
from datasets import load_dataset
d = load_dataset("brimmann2/squad-v2-sampled")
```

```python
generated_answers = []
```

```python
for emb in d["train"]["embeddings"]:
    emb_tensor = torch.tensor(emb).to(device)
    output = model.generate(
        **tokenized_prompt,
        do_sample=False,
        max_new_tokens=200,
        retrieval_embeds = emb_tensor
    )
    generated_answers.append(tokenizer.decode(output[0]))
```
