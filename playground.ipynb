```python
from transformers import AutoTokenizer
from imported_code.modeling_xmistral import XMistralForCausalLM
from datasets import load_from_disk
```

```python
import torch
```

```python
XRAG_TOKEN = "<xRAG>" 
```

```python
device = "cuda" if torch.cuda.is_available() else "cpu"
```

```python
model_name = "Hannibal046/xrag-7b"
model = XMistralForCausalLM.from_pretrained(
    model_name,
    torch_dtype = torch.bfloat16,
    low_cpu_mem_usage = True,
    )
model.to(device)
tokenizer = AutoTokenizer.from_pretrained(
    model_name,
    add_eos_token=False,
    use_fast=False,
    padding_side='left'
    )
model.set_xrag_token_id(tokenizer.convert_tokens_to_ids(XRAG_TOKEN))
```

```python
rag_template = """[INST] Background: {xrag_token}, which also means: [/INST]"""
prompt = rag_template.format_map(dict(xrag_token=XRAG_TOKEN))
print(prompt)
```

```python
tokenized_prompt = tokenizer(prompt, return_tensors="pt").to(device)
```

```python
from datasets import load_dataset
d = load_dataset("brimmann2/squad-v2-sampled")
```

```python
batch_size = 64
embeddings = d["train"]["embeddings"]
generated_answers = []
```

```python
from tqdm import tqdm
```

```python
for i in tqdm(range(0, len(embeddings), batch_size), desc="Generating answers:"):
    batch_embs_list = embeddings[i:i+batch_size]

    emb_tensor = torch.tensor(batch_embs_list).to(device)
    current_batch_size = emb_tensor.shape[0]

    batch_prompt = {
        key: value.expand(current_batch_size, -1)
        for key, value in tokenized_prompt.items()
    }

    output = model.generate(
        **batch_prompt,
        do_sample=False,
        max_new_tokens=200,
        retrieval_embeds=emb_tensor
    )
```
